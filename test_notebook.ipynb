```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
with open("data\\train-consonant-examples.txt", 'r') as text:
    data = text.readlines()
```

```python
print(data)
cleaned_data = []
for sentence in data:
    cleaned_data.append(sentence.strip('\n'))
```

```python
word_list = []
for i,sentence in enumerate(cleaned_data):
    words = sentence.split(" ")
    word_list.extend(words)
```

```python
len_list = []
for word in word_list:
    len_list.append(len(word))
```

```python
plt.figure(figsize=(10, 6))
plt.hist(len_list, bins=range(1, max(len_list)+2), edgecolor='black')
plt.xlabel('Word Length')
plt.ylabel('Frequency')
plt.title('Histogram of Word Lengths in Training Data')
plt.grid(True)
plt.xticks(range(1, max(len_list)+1))  
plt.show()
```
